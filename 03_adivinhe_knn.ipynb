Este notebook usa K-Nearest Neighbors para adivinhar o personagem.


```python
import pandas
import matplotlib.pyplot as plt
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import confusion_matrix, accuracy_score, classification_report
```

```python
dados = pandas.read_csv("personagens.csv")

X = dados[["magia", "forca", "inteligencia"]]
y = dados["personagem"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.5, random_state=42
)
```

```python
model = KNeighborsClassifier(n_neighbors=3)
model.fit(X_train, y_train)

y_pred = model.predict(X_test)
```

```python
acc = accuracy_score(y_test, y_pred)
print(f"Acurácia (KNN): {acc:.2f}\n")
print("Relatório de classificação:\n")
print(classification_report(y_test, y_pred))
```

```python
from sklearn.metrics import confusion_matrix
import numpy as np
import matplotlib.pyplot as plt

cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(14, 10))
plt.imshow(cm, interpolation='nearest')
plt.title('Matriz de Confusão - KNN', fontsize=16)
plt.colorbar()

# Classes presentes apenas no conjunto de teste
classes_teste = np.unique(y_test)

tick_marks = np.arange(len(classes_teste))
plt.xticks(tick_marks, classes_teste, rotation=90, fontsize=9)
plt.yticks(tick_marks, classes_teste, fontsize=9)

# Números dentro das células
for i in range(cm.shape[0]):
    for j in range(cm.shape[1]):
        plt.text(j, i, cm[i, j],
                 ha="center", va="center",
                 color="black" if cm[i, j] < cm.max()/2 else "white",
                 fontsize=8)

plt.xlabel("Previsto", fontsize=12)
plt.ylabel("Real", fontsize=12)
plt.tight_layout()
plt.show()
```

# Conclusão - KNN

Resumo: KNN classifica com base nos personagens mais parecidos no conjunto de treino.  
Interpretação simples: comparar acurácia e ver qual modelo teve menos confusões (matriz).
