Este notebook usa Random Forest para adivinhar o personagem dada as características: magia, forca e inteligencia.


```python
import pandas
import matplotlib.pyplot as plt
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import confusion_matrix, accuracy_score, classification_report
```

```python
dados = pandas.read_csv("personagens.csv")

X = dados[["magia", "forca", "inteligencia"]]
y = dados["personagem"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.5, random_state=42
)
```

```python
model = RandomForestClassifier()
model.fit(X_train, y_train)

y_pred = model.predict(X_test)
```

```python
acc = accuracy_score(y_test, y_pred)
print(f"Acurácia (Random Forest): {acc:.2f}\n")
print("Relatório de classificação:\n")
print(classification_report(y_test, y_pred))
```

```python
cm = confusion_matrix(y_test, y_pred)


plt.figure(figsize=(14, 10))
plt.imshow(cm, interpolation='nearest', cmap='Blues')   # ALTERE AQUI A COR
plt.title('Matriz de Confusão - Random Forest', fontsize=16)
plt.colorbar()

tick_marks = np.arange(len(model.classes_))
plt.xticks(tick_marks, model.classes_, rotation=90, fontsize=9)
plt.yticks(tick_marks, model.classes_, fontsize=9)

for i in range(cm.shape[0]):
    for j in range(cm.shape[1]):
        plt.text(j, i, cm[i, j],
                 ha="center", va="center",
                 color="black" if cm[i, j] < cm.max()/2 else "white",
                 fontsize=8)

plt.xlabel('Previsto', fontsize=12)
plt.ylabel('Real', fontsize=12)
plt.tight_layout()
plt.show()
```
